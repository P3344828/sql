**Q1.** Explain the fundamental differences between DDL, DML, and DQL commands in SQL. Provide one example for each type of command.

- **Ans:** **DDL, DML, and DQL** are fundamental categories of SQL commands, each serving a distinct function in database management.

- ** Data Definition Language (DDL):** Data Definition Language (DDL) is used to define, modify, and manage the structure of database objects such as tables, schemas, and indexes.
  - Function: Defines and alters database structure
  - Examples of Commands: `CREATE`, `ALTER`, `DROP`, `TRUNCATE`
  - Example:

```sql
CREATE TABLE Students (
    ID INT,
    Name VARCHAR(50),
    Age INT
);
```
- **Data Manipulation Language (DML):** Data Manipulation Language (DML) is used to manage and manipulate data stored within database tables.
  - Function: Inserts, updates, and deletes data
  - Examples of Commands: `INSERT`, `UPDATE`, `DELETE`
  - Example:

```sql
INSERT INTO Students (ID, Name, Age)
VALUES (1, 'Priya', 20);
```
- **Data Query Language (DQL):** Data Query Language (DQL) is used to retrieve data from the database.
  - Function: Queries and fetches data
  - Primary Command: `SELECT`
  - Example:

```sql
SELECT * FROM Students;
```

**Q2.** What is the purpose of SQL constraints? Name and describe three common types of constraints, providing a simple scenario where each would be useful.

- **Ans:** SQL constraints are rules applied to columns in a database table to ensure data integrity, accuracy, and consistency. They restrict the type of data that can be inserted or modified, thereby preventing invalid or inconsistent data from being stored.

- **PRIMARY KEY Constraint:** The PRIMARY KEY constraint uniquely identifies each record in a table. It ensures that the column contains only unique and non-null values.
  - Purpose: To maintain entity integrity by uniquely identifying each row
  - Scenario: In a *Students* table, each student must have a unique  `Student_ID` to distinguish records.
  - Example:

```sql id="q9k2zt"
CREATE TABLE Students (
    Student_ID INT PRIMARY KEY,
    Name VARCHAR(50)
);
```
- **NOT NULL Constraint:** The NOT NULL constraint ensures that a column cannot store NULL values, meaning every record must contain a value for that column.
  - Purpose: To enforce mandatory data entry for essential attributes
  - Scenario: In an *Employees* table, the  `Name` field must always be provided for proper identification.
  - Example:

```sql id="mjf6vb"
CREATE TABLE Employees (
    ID INT,
    Name VARCHAR(50) NOT NULL
);
```
- **FOREIGN KEY Constraint:** The FOREIGN KEY constraint is used to establish a relationship between two tables by linking a column in one table to the PRIMARY KEY of another table.
  - Purpose: To maintain referential integrity between related tables
  - Scenario: In an *Orders* table, the  `Customer_ID` must refer to an existing record in the *Customers* table.
  - Example:

```sql id="m4n1qg"
CREATE TABLE Orders (
    Order_ID INT,
    Customer_ID INT,
    FOREIGN KEY (Customer_ID) REFERENCES Customers(Customer_ID)
);
```

**Q3.** Explain the difference between LIMIT and OFFSET clauses in SQL. How would you use them together to retrieve the third page of results, assuming each page has 10 records?

- **Ans:** The **LIMIT** and **OFFSET** clauses in SQL are used to control the number of rows returned by a query, particularly for implementing pagination in result sets.

- **LIMIT Clause:** The LIMIT clause is used to specify the maximum number of records to be retrieved from a query result.
  - Function: Restricts the number of rows returned
  - Example:

```sql id="l2z8df"
SELECT * FROM Students
LIMIT 10;
```
This query returns only the first 10 records from the *Students* table.

- **OFFSET Clause:** The OFFSET clause is used to specify the number of rows to skip before beginning to return records.
  - Function: Skips a specified number of rows
  - Example:

```sql id="y6kp9x"
SELECT * FROM Students
OFFSET 5;
```
This query skips the first 5 records and returns the remaining rows.

- **Using LIMIT and OFFSET Together:** When combined, the LIMIT and OFFSET clauses facilitate efficient pagination by retrieving a specific subset of records from a larger dataset.
  - Formula:

  OFFSET = (Page Number − 1) × Number of Records per Page

- **Retrieving the Third Page (10 Records per Page):**
  - For the third page, where each page contains 10 records:
    - Page Number = 3
    - Records per Page = 10
    - OFFSET = (3 − 1) × 10 = 20
  - Query:

```sql id="xg4t9v"
SELECT * FROM Students
LIMIT 10 OFFSET 20;
```

**Q4.** What is a Common Table Expression (CTE) in SQL, and what are its main benefits? Provide a simple SQL example demonstrating its usage.

- **Ans:** A **Common Table Expression (CTE)** in SQL is a temporary, named result set that is defined within the scope of a single query using the `WITH` clause. It can be referenced within a `SELECT`, `INSERT`, `UPDATE`, or `DELETE` statement, and exists only for the duration of that query.

- **Purpose of a CTE:** The primary purpose of a CTE is to simplify complex SQL queries by dividing them into smaller, logical components, thereby improving clarity and maintainability.

- **Main Benefits of a CTE**

  - **Enhanced Readability:** CTEs allows complex queries to be structured in a clear and organized manner, making them easier to understand.

  - **Improved Query Maintainability:** By breaking down complex logic into smaller parts, modifications and debugging become more manageable.

  - **Reusability Within a Query:** A CTE can be referenced multiple times within the same query, reducing redundancy.

  - **Support for Recursive Operations:** CTEs enable recursive queries, which are useful for handling hierarchical data such as organizational structures or tree-based relationships.

- **Example**

```sql id="d7w3qx"
WITH HighSalaryEmployees AS (
    SELECT Name, Salary
    FROM Employees
    WHERE Salary > 50000
)
SELECT *
FROM HighSalaryEmployees;
```
- **Explanation:** In this example, a CTE named `HighSalaryEmployees` is defined to store the result of a query that selects employees with salaries greater than 50,000. The main query then retrieves data from this CTE as though it were a standard table.

**Q5.** Describe the concept of SQL Normalization and its primary goals. Briefly explain the first three normal forms (1NF, 2NF, 3NF).

- **Ans:** **SQL Normalization** is a systematic approach used in database design to organize data in a structured and efficient manner. It involves decomposing large tables into smaller, well-structured tables to minimize redundancy and eliminate undesirable characteristics such as insertion, update, and deletion anomalies.

- ** Primary Goals of Normalization**

  - **Reduction of Data Redundancy:** Ensures that duplicate data is minimized across the datbase.
  
  - **Enhancement of Data Integrity:** Maintains accuracy and consistency of data through well-defined relationships.

  - **Elimination of Anomalies:** Prevents issues during data insertion, updating, and deletion.

  - ** Improved Data Organization:** Structures data logically to facilitate efficient storage and retrieval.

- **First Three Normal Forms**
  - **First Normal Form (1NF):** A relation is said to be in First Normal Form if:
     - All attributes contain atomic (indivisible) values
     - There are no repeating groups or multivalued attributes
     - Each record can be uniquely identified

   - Explanation: Data is organized so that each field contains only a single value, ensuring a tabular structure with no nested or grouped data.

  - **Second Normal Form (2NF):** A relation is in Second Normal Form if:
     - It is already in 1NF
     - All non-key attributes are fully functionally dependent on the entire primary key
   - Explanation: Partial dependencies are removed, meaning non-key attributes must depend on the whole primary key rather than a part of it (especially in the case of composite keys).
  
  - **Third Normal Form (3NF):** A relation is in Third Normal Form if:
     - It is already in 2NF
     - There are no transitive dependencies between non-key attributes
   - Explanation: Non-key attributes should depend only on the primary key and  not on other non-key attributes, ensuring further reduction of redundancy.

**Q7.** Generate a report showing CustomerName, Email, and the TotalNumberofOrders for each customer. Include customers who have not placed
any orders, in which case their TotalNumberofOrders should be 0. Order the results by CustomerName.


````python
import pandas as pd
import sqlite3

# Re-establish the database connection for this cell
conn = sqlite3.connect('ECommerceDB.db')

query = """
SELECT
    Customers.CustomerName,
    Customers.Email,
    COUNT(Orders.OrderID) AS TotalNumberOfOrders
FROM Customers
LEFT JOIN Orders
ON Customers.CustomerID = Orders.CustomerID
GROUP BY Customers.CustomerID,
         Customers.CustomerName,
         Customers.Email
ORDER BY Customers.CustomerName;
"""

df = pd.read_sql_query(query, conn)

print(df)

# Close the connection after use in this cell
conn.close()
````

**Q8.** Retrieve Product Information with Category: Write a SQL query to

products. Order the results by CategoryName and then ProductName alphabetically.

````python
import pandas as pd
import sqlite3

# Re-establish the database connection for this cell
conn = sqlite3.connect('ECommerceDB.db')

query = """
SELECT
    Products.ProductName,
    Products.Price,
    Products.StockQuantity,
    Categories.CategoryName
FROM Products
INNER JOIN Categories
ON Products.CategoryID = Categories.CategoryID
ORDER BY Categories.CategoryName,
         Products.ProductName;
"""

df = pd.read_sql_query(query, conn)

print(df)

# Close the connection after use in this cell
conn.close()
````

**Q9.** Write a SQL query that uses a Common Table Expression (CTE) and a
Window Function (specifically ROW_NUMBER() or RANK()) to display the
CategoryName, ProductName, and Price for the top 2 most expensive products in
each CategoryName.

````python
import pandas as pd
import sqlite3

# Re-establish the database connection for this cell
conn = sqlite3.connect('ECommerceDB.db')

query = """
WITH ProductRank AS (
    SELECT
        Categories.CategoryName,
        Products.ProductName,
        Products.Price,

        ROW_NUMBER() OVER (
            PARTITION BY Categories.CategoryName
            ORDER BY Products.Price DESC
        ) AS RankNum

    FROM Products

    INNER JOIN Categories
    ON Products.CategoryID = Categories.CategoryID
)

SELECT
    CategoryName,
    ProductName,
    Price
FROM ProductRank

WHERE RankNum <= 2

ORDER BY CategoryName, Price DESC;
"""

# Pandas DataFrame
df = pd.read_sql_query(query, conn)

# Output
print(df)

# Close the connection after use in this cell
conn.close()
````

**Q10.** You are hired as a data analyst by Sakila Video Rentals, a global movie rental company. The management team is looking to improve decision-making by analyzing existing customer, rental, and inventory data.

Using the Sakila database, answer the following business questions to support key strategic
initiatives.

**Tasks & Questions:**
  - 1. Identify the top 5 customers based on the total amount they’ve spent. Include customer
name, email, and total amount spent.
  - 2. Which **3 movie categories** have the **highest rental counts**? Display the category name
and number of times movies from that category were rented.
 - 3. Calculate how many films are available at each store and how many of those have
**never been rented**.
 - 4. Show the **total revenue per month** for the year 2023 to analyze business seasonality.
 - 5. Identify customers who have rented **more than 10 times** in the last 6 months.

````python
import sqlite3
import pandas as pd
import os
import requests # Import the requests library

# Connect to sakila.db
conn = sqlite3.connect("sakila.db")

# List of Sakila CSV files
sakila_csv_files = [
    'Sakila.customer.csv',
    'Sakila.payment.csv',
    'Sakila.rental.csv',
    'Sakila.category.csv',
    'Sakila.film_category.csv',
    'Sakila.inventory.csv',
    'Sakila.store.csv',
    'Sakila.film.csv'
]

# Base URL for downloading the CSVs from GitHub
base_url = "https://raw.githubusercontent.com/jayasuryamallam/Sakila-Database/main/"

print("Attempting to download Sakila CSV files using requests...")
for csv_file in sakila_csv_files:
    if not os.path.exists(csv_file): # Check if file already exists locally
        file_url = base_url + csv_file
        try:
            response = requests.get(file_url)
            response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
            with open(csv_file, 'wb') as f:
                f.write(response.content)
            print(f"Downloaded {csv_file}")
        except requests.exceptions.RequestException as e:
            print(f"Could not download {csv_file} from {file_url}: {e}")
            print(f"Please ensure '{csv_file}' is uploaded to your Colab environment or the URL is correct.")
    else:
        print(f"{csv_file} already exists locally.")

print("\nLoading data into sakila.db...")
# Load each CSV into a DataFrame and then into SQLite
for csv_file in sakila_csv_files:
    if os.path.exists(csv_file):
        try:
            # Use sep=';' to handle semicolon-delimited CSVs
            df = pd.read_csv(csv_file, sep=';')
            # Sanitize table name: remove 'Sakila.' prefix and replace '.' with '_'
            table_name = csv_file.replace('Sakila.', '').replace('.csv', '').replace('.', '_')
            df.to_sql(table_name, conn, if_exists='replace', index=False)
            print(f"Loaded {csv_file} into table: {table_name}")
        except Exception as e:
            print(f"Error loading {csv_file} into SQLite: {e}")
    else:
        print(f"Warning: {csv_file} not found locally. Skipping. Please ensure it's available.")

# Verify tables are created
print("\nTables in sakila.db after loading CSVs:")
display(pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn))

conn.close()
````

````python
# 1. Top 5 customer
import sqlite3
import pandas as pd

# Connect to sakila.db
conn = sqlite3.connect('sakila.db')

query1 = """
SELECT
    c.first_name || ' ' || c.last_name AS CustomerName,
    c.email,
    SUM(p.amount) AS TotalAmountSpent
FROM customer c
JOIN payment p
ON c.customer_id = p.customer_id
GROUP BY c.customer_id, c.first_name, c.last_name, c.email
ORDER BY TotalAmountSpent DESC
LIMIT 5;
"""

df1 = pd.read_sql_query(query1, conn)

print("Top 5 Customers by Total Amount Spent:")
display(df1)

# Close the connection
conn.close()
````

````python
# 2.  3 movie categories have the highest rental counts
import sqlite3
import pandas as pd

# Re-establish the database connection for this cell
conn = sqlite3.connect('sakila.db')

query2 = """
SELECT
    c.name AS CategoryName,
    COUNT(r.rental_id) AS RentalCount
FROM category c
JOIN film_category fc ON c.category_id = fc.category_id
JOIN film f ON fc.film_id = f.film_id
JOIN inventory i ON f.film_id = i.film_id
JOIN rental r ON i.inventory_id = r.inventory_id
GROUP BY c.name
ORDER BY RentalCount DESC
LIMIT 3;
"""

try:
    df2 = pd.read_sql_query(query2, conn)
    print("Top 3 Movie Categories by Rental Count:")
    display(df2)
except pd.io.sql.DatabaseError as e:
    print(f"Error executing query for: {e}")
    print("This error likely indicates that one or more required tables are missing from sakila.db.")

# Close the connection
conn.close()
````

````python
# 3.  films are available at each store and how many of those have never been rented.
import sqlite3
import pandas as pd

# Re-establish the database connection for this cell
conn = sqlite3.connect('sakila.db')

query3 = """
SELECT
    s.store_id,
    s.address_id AS StoreAddressId,
    COUNT(DISTINCT i.film_id) AS TotalFilmsAvailable,
    COUNT(DISTINCT CASE WHEN r.rental_id IS NULL THEN i.film_id ELSE NULL END) AS FilmsNeverRented
FROM store s
LEFT JOIN inventory i ON s.store_id = i.store_id
LEFT JOIN rental r ON i.inventory_id = r.inventory_id
GROUP BY s.store_id, s.address_id -- Group by identifying store info
ORDER BY s.store_id;
"""

try:
    df3 = pd.read_sql_query(query3, conn)
    print("Films available at each store and films never rented:")
    display(df3)
except pd.io.sql.DatabaseError as e:
    print(f"Error executing query for: {e}")
    print("This error likely indicates that one or more required tables are missing from sakila.db.")

# Close the connection
conn.close()
````

````python
# 4. Total revenue per month
import sqlite3
import pandas as pd

# Re-establish the database connection for this cell
conn = sqlite3.connect('sakila.db')

query4 = """
SELECT
    strftime('%Y-%m', payment_date) AS SaleMonth,
    SUM(amount) AS TotalRevenue
FROM payment
WHERE strftime('%Y', payment_date) = '2005' -- Filter for the year 2005
GROUP BY SaleMonth
ORDER BY SaleMonth;
"""

df4 = pd.read_sql_query(query4, conn)

print("Total Revenue per Month (2005):")
display(df4)

# Close the connection
conn.close()
````

````python
# 5. Customers with more than 10 rentals
import sqlite3
import pandas as pd

# Re-establish the database connection for this cell
conn = sqlite3.connect('sakila.db')

query5 = """
SELECT
    c.first_name || ' ' || c.last_name AS CustomerName,
    c.email,
    COUNT(r.rental_id) AS RentalCount
FROM customer c
JOIN rental r ON c.customer_id = r.customer_id
WHERE r.rental_date >= '2005-02-01'     -- Filter for the last 6 months
GROUP BY c.customer_id, c.first_name, c.last_name, c.email
HAVING COUNT(r.rental_id) > 10
ORDER BY RentalCount DESC;
"""

df5 = pd.read_sql_query(query5, conn)

print("Customers with more than 10 rentals in a relevant 6-month period:")
display(df5.head(10))

# Close the connection
conn.close()
````
